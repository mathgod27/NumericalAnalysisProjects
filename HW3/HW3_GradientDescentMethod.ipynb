## HW 3 Task 2 is to minimize the function using Gradient Descent and Newton's method.

```python
# import the libraries
import numpy as np
import math
```

## Example from Class for implementing the gradient vector

gradient1.PNG

For this problem, our equation is different from the example and the equation we are using for the following code is f(x) = ($x_{1}$ - 1)^2 + ($x_{1}$^2 - $x_{2}$)^2

```python
## a function returns the Gradient vector at the input vector x_i

def getGradientVector(x_i):
  g = np.zeros((2,1))   #changed because we only have 2 x's
  g[0] = 2 * (x_i[0] - 1) + 2*2*x_i[0]*(x_i[0]**2 - x_i[1])
  g[1] = -2 * (x_i[0]**2 - x_i[1])
  return g

# a simple function returns the norm of an input vector
def getNorm(v):
  return math.sqrt(v[0]*v[0]+v[1]*v[1])   #again, only 2 x's
```

## Example of Performing the Gradient Descent Method for minimizing the nonlinear function

gradient2.PNG

Our initial guess array for this problem is as follows:  
$x_{0}$ = [3; 3]  
Threshold = 0.01  
Alpha = 0.01  
because we don't want our alpha to be too large

```python
## set an initial guess
x_0 = np.array([[3], [3]])
print(f"--The initial guess x0: ")
print(x_0)

## set a threshold value for getting a good solution
thresh = 0.01

## set step length
alpha = 0.01

x_i = x_0
count = 0
while True:
  print('****************************************************')
  print(f'--At iteration: {count}')

  ## compute the gradient vector at x_i
  g = getGradientVector(x_i)
  print(f"--The gradient vector g(x_{count}): ")
  print(g)

  ## compute the norm of the gradient vector at x_i
  normOfGradient = getNorm(g)
  print(f"--The norm of the gradient vector |g(x_{count})|: {normOfGradient}")

  ## check if converged; if true, then stop; continue otherwise
  if normOfGradient <= thresh:
    break


  ## get h=-g
  h_i = -1*g
  print(f"--The h vector h_{count}: ")
  print(h_i)

  ## get the new solution point
  x_i_plus1 = x_i + alpha * h_i
  print(f"--The solution vector x_{count+1}: ")
  print(x_i_plus1)

  x_i = x_i_plus1
  count = count + 1
  print()
```

Our number of iterations used to reach convergence is 1556  
Our new values of $x_{1}$ and $x_{2}$ are $x_{1}$ = [1.01124206] and $x_{2}$ = [1.02722925]  
Our function value at convergence is calculated by plugging in the values to our original equation.  
f($x_{1}$,$x_{2}$) = 0.0001478

The conclusions we can make from comparing both methods is that Newton's Method converged in less than 5 iterations, however gradient descent is simpler, more scalable, and more practical.  
If you want for minimizing your function to be a small amount of iterations and don't mind heavier work and needing to be more specific, choose Newton's Method.  
If you want for minimizing your function and working effectively without a good initial guess, choose Gradient Descent method.
